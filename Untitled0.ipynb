```python
def industry_group(code):
    if pd.isna(code):
        return "Unknown"

    code = int(code)
    division = code // 1000

    if 1 <= division <= 3:
        return "Agriculture"

    elif 5 <= division <= 43:
        return "Industry"

    elif 45 <= division <= 99:
        return "Services"

    else:
        return "Other"


working_df["Industry_Group"] = (
    working_df["Principal_Industry_Code"].apply(industry_group)
)

working_df["Industry_Group"].value_counts()
```

```python
# ML dataset
ml_df = working_df[
    [
        "Avg_Positive_Daily_Wage",
        "Avg_Daily_Hours",
        "General_Education_Level",
        "Principal_Occupation_Code",
        "Industry_Group"
    ]
].copy()

# Only valid wage + hours observations
ml_df = ml_df.dropna(
    subset=["Avg_Positive_Daily_Wage", "Avg_Daily_Hours"]
)

print("ML dataset shape:", ml_df.shape)
print(ml_df.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Features
X = ml_df[
    [
        "Avg_Daily_Hours",
        "General_Education_Level",
        "Principal_Occupation_Code",
        "Industry_Group"
    ]
]

# Target
y = ml_df["Avg_Positive_Daily_Wage"]

# Convert categorical variables to strings
X = X.copy()
X["General_Education_Level"] = X["General_Education_Level"].astype(str)
X["Principal_Occupation_Code"] = X["Principal_Occupation_Code"].astype(str)
X["Industry_Group"] = X["Industry_Group"].astype(str)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# Categorical + numerical columns
categorical_cols = [
    "General_Education_Level",
    "Principal_Occupation_Code",
    "Industry_Group"
]

numeric_cols = ["Avg_Daily_Hours"]

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
        ("num", "passthrough", numeric_cols)
    ]
)

# Random Forest
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

# Train
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Evaluate
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R² Score:", round(r2, 3))
```

```python
# Get feature importance from Random Forest
rf_model = model.named_steps["model"]
preprocessor_fitted = model.named_steps["preprocessor"]

feature_names = preprocessor_fitted.get_feature_names_out()

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": rf_model.feature_importances_
}).sort_values("Importance", ascending=False)

print(importance_df.head(20).to_string(index=False))
```
